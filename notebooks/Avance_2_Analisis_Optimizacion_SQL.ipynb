## Optimización SQL con índices

### Comparación de rendimiento

Consulta analizada: conteo de entregas realizadas durante los últimos 90 días.

- Tiempo antes de optimizar: 253,653 ms.
- Tiempo después de optimizar: 36,815 ms.
- Reducción del tiempo: 85,48 %.
- Aceleración aproximada: 6,89 veces.

### Índice implementado

CREATE INDEX IF NOT EXISTS idx_deliveries_status_scheduled
ON deliveries (delivery_status, scheduled_datetime);

### Justificación técnica

Se creó un índice compuesto sobre las columnas
delivery_status y scheduled_datetime para facilitar
el filtrado por estado y fecha programada.

Las mediciones con EXPLAIN ANALYZE muestran una
reducción significativa del tiempo de ejecución.

### Impacto empresarial

La optimización permite consultar más rápidamente
las entregas recientes, facilitando el monitoreo
operativo y la elaboración de reportes logísticos.
